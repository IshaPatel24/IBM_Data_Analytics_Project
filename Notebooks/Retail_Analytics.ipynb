# IBM Data Analytics Project

## Project Overview

This project performs end-to-end data analytics on the Online Retail dataset.

## Technologies

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Scikit-Learn
- Jupyter Notebook

## Project Workflow

1. Data Cleaning
2. Exploratory Data Analysis
3. Feature Engineering
4. Revenue Analysis
5. Product Analysis
6. Country Analysis
7. Monthly Revenue Trend
8. RFM Analysis
9. K-Means Customer Segmentation

## Results

- Top Products
- Top Countries
- Monthly Revenue
- Customer Segments
